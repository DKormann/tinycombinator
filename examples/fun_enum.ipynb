```python
import numpy as np
from tinycombinator.term import Term
import random
from matplotlib import pyplot as plt
```

```python
x = np.arange(1000)
```

```python
x[-10:]
```

```python
uops = ['-', 'not']
binops = ['+', '-', '*', '%']
tertops = ['if else']


def sample_r(d):
  if d == 0: raise ValueError('d must be at least 1')
  if d == 1 :
    if random.random() > 0.5:
      return str(int(2**(random.random()*3)))
    return 'x'
  
  d1 = int(random.random() * (d -1)) + 1
  d2 = d - d1

  op = random.choice(binops)

  return f'({sample_r(d1)} {op} {sample_r(d2)})'

def sample_fun(d):
  code = f'lambda x: 0<{sample_r(d)}'
  return code, eval(code)


sample_fun(5)
  
```

```python

for i in range(10):
  c,f = sample_fun(5)
  T = f(x)

  s = T.sum()

  if s < 10 or s > len(x) - 10:
    print("reject.")
    continue
  break

plt.plot(T)
```
