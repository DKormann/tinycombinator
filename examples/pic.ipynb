```python
import numpy as np
import random
```

```python
RED = "\033[91m"
GREEN = "\033[92m"
BLUE = "\033[94m"
RESET = "\033[0m"


def print_color(color, text):
  print(f"{color}{text}\033[0m")

```

```python
from dataclasses import dataclass

@dataclass
class Token:
  val: int
  color: str


field = [[None for _ in range(4)] for _ in range(4)]

field[0][0] = Token(1, RED)
field[0][1] = Token(2, GREEN)
field[0][2] = Token(3, BLUE)
field[0][3] = Token(4, RED)


def show(field):
  for row in field:
    print(" ".join(f"{t.color}{t.val}" if t else ' ' for t in row))

show(field)
```

```python
def col(x): return x % 3
def val(x): return x // 3



def new_field():
  return np.zeros((4,4), dtype=int)

def set_token(field, x,y,val, col):
  assert 0 < val < 10
  assert col < 3
  field[x,y] = val * 3 + col

def del_token(field, x,y):
  field[x,y] = 0

def show_field(field):
  for row in field:
    print(" ".join(f"{[RED, GREEN, BLUE][col(x)]}{val(x)}" if x else ' ' for x in row))

field = np.zeros((4,4), dtype=int)
set_token(field, 0,0,1,0)
show_field(field)
```

```python
field = new_field()
for i in range(6):
  set_token(field,
    random.randint(0,3),
    random.randint(0,3),
    random.randint(1,9),
    random.randint(0,2)
  )
show_field(field)

```

```python
def empty(field):
  return field == 0

def is_color(field, col):
  return (field > 0) * (field % 3 == col)


def top_of(T): return np.pad(T[1:], ((0,1), (0,0)), mode='constant')
def left_of(T): return np.pad(T[:,1:], ((0,0), (0,1)), mode='constant')
def right_of(T): return np.pad(T[:,:-1], ((0,0), (0,1)), mode='constant')
def bottom_of(T): return np.pad(T[:-1], ((0,1), (0,0)), mode='constant')
    
    

    

def view_rule(field, rule):
  T = rule(field)
  for xrow, trow in zip(field, T):
    print(" ".join(f'{"x" if t else "o"}{[RED, GREEN, BLUE][col(x)]}{val(x)}{RESET}' if x else '  ' for x,t in zip(xrow, trow)))


view_rule(field, lambda x: top_of(is_color(x, 2)))


```
